Depth-first traversal

```python
class Node:
    def __init__(self, val, left=None, right=None):
        self.val, self.left, self.right = val, left, right

# Recursive
def dfs_rec(node, out):
    if node is None:
        return
    out.append(node.val)
    dfs_rec(node.left, out)
    dfs_rec(node.right, out)

# Iterative (explicit stack)
def dfs_iter(root):
    out, stack = [], [root] if root else []
    while stack:
        node = stack.pop()
        out.append(node.val)
        if node.right: stack.append(node.right)
        if node.left:  stack.append(node.left)
    return out
```

Binary-Search - degrades in space

```python
# Iterative: O(log n) time, O(1) space
def bsearch_iter(a, x):
    lo, hi = 0, len(a) - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        if a[mid] == x: return mid
        if a[mid] < x: lo = mid + 1
        else:          hi = mid - 1
    return -1

# Recursive: O(log n) time, O(log n) space (stack frames)
def bsearch_rec(a, x, lo, hi):
    if lo > hi: return -1
    mid = (lo + hi) // 2
    if a[mid] == x: return mid
    if a[mid] < x: return bsearch_rec(a, x, mid + 1, hi)
    return bsearch_rec(a, x, lo, mid - 1)
```

Fibonacci - degrades in time

```python
# Iterative: O(n) time, O(1) space
def fib_iter(n):
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a

# Naive recursive: exponential time, O(n) space
def fib_rec(n):
    if n < 2: return n
    return fib_rec(n - 1) + fib_rec(n - 2)
```
